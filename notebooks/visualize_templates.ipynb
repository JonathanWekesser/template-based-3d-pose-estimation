# Visualize Templates

```python
import os
import sys

import open3d as o3d
from open3d.camera import PinholeCameraIntrinsic, PinholeCameraParameters

sys.path.append(os.path.abspath("../src"))
from paths import PathManager
from template import Template, TemplateLoader
```

```python
def show_template(pcd, title="Template", plotly=False):
    o3d.visualization.draw_geometries([pcd], window_name=title)
```

## apples

```python
item = "apple"
pm = PathManager(item)
loader = TemplateLoader(pm)
templates = loader.load_templates()

print(f"{len(templates)} Templates geladen:")
for template in templates:
    print(" -", template.name)
    show_template(template.pcd)
```
